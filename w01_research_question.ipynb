## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Looking at the dataset statistics, the numbers are making it very clear why this problem is actually worth solving:

Out of 30,000 pages that are older than 90 days, 16,262 pages (54.2%) are currently losing traffic. Even worse, 9,961 of these dropping pages still get at least 500 impressions. This means people are actively searching for these topics, but our articles are failing to bring them in so there must be a dire need of fixing on the article/editor's end.

Manually sifting through almost 10,000 dying pages is impossible for a human team. These numbers show that a huge portion of search traffic is slipping away, which again proves we need an automated system to catch and prioritize them.

So spending time, effort and focus here is purposeful and justified!

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os
import pandas as pd

if not os.path.exists("data"):
    !git clone https://github.com/musfira-tahir-ai/flyrank-assignment-1.git
    %cd flyrank-assignment-1

#loading the dataset
data_path = "../../data/raw/content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    data_path = "data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

#selecting pages above 90 days age so we can easily judge them!
filtered_df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)].copy()

total_pages = len(filtered_df)
declining_pages = (filtered_df["trend_direction"] == "down").sum()
decline_pct = (declining_pages / total_pages) * 100

#comparing impressions vs clicks or traffic
high_demand_declining = (
    (filtered_df["trend_direction"] == "down") &
    (filtered_df["impressions_90d"] >= 500)
).sum()

print(f"Total mature pages checked: {total_pages:,}")
print(f"Pages with declining traffic: {declining_pages:,} ({decline_pct:.1f}%)")
print(f"High-demand declining pages (>= 500 impressions): {high_demand_declining:,}")
```
